```python
import altair as alt
plot_2 = alt.Chart(df).mark_bar().encode(
    alt.X("Conference:N",sort="-y", title="Conference"),
    alt.Y("count()", title="Count")
).properties(
    title="Ralative Proportion of Teams per Conference",
    width=500,
    height=300
)
plot_2
```

Visualization 1: This box plot examines the relationship between a team's total wins and whether they advanced to the final four. It allows us to compare the median and spread of total wins between the two groups, helping us evaluate if total wins is avstrong predictor for tournament success.

```python
import altair as alt
vis_1 = alt.Chart(df).mark_boxplot().encode(
    alt.X("made_final_four:N", title="Made final Four(True/False)"),
    alt.Y("total_wins:Q", title="Total Wins")
).properties(
    title="Total Wins Distribution by Final Four Status",
    width=450,
    height=300
)
vis_1
```

Visualization 2: This box plot explores how a team's seed ranking relates to their qualification for the final four. By comparing the seed distributions, we can visually assess whether higher seeded teams are more likely to reach the final compete.

```python
import altair as alt
vis_2 = alt.Chart(df).mark_boxplot().encode(
    alt.X("made_final_four:N", title="Made final Four(True/False)"),
    alt.Y("seed:Q", title="Team Seed(Lower is better)")
).properties(
    title="Seed Distribution by Final Four Status",
    width=450,
    height=300
)
vis_2
```
